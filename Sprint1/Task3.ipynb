```
public class Movie
{
    public string Title {get; set;}
    public string Director {get; set;}
    public int ReleaseYear {get; set;}

    public Movie(string title, string director, int releaseyear)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseyear;
    }

    public virtual string GetInfo()
    {
        return $"{Title}, {Director}, {ReleaseYear}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"Вы смотрите фильм: {Title}");
    }

    public virtual void Rate(int rating)
    {
        Console.WriteLine($"{Title} оценен на {rating}/10");
    }
}   

public class Documentary : Movie
{
    private string Theme {get; set;}

    public Documentary(string title, string director, int releaseyear, string theme) 
    : base (title, director, releaseyear)
    {
        Theme = theme;
    }

    public override void Watch()
    {
        Console.WriteLine($"Вы смотрите фильм {Title} на тему {Theme}");
    }
}

public class FeatureFilm : Movie
{
    private string Genre {get; set;}

    public FeatureFilm(string title, string director, int releaseyear, string genre)
    : base (title, director, releaseyear)
    {
        Genre = genre;
    }

    public override void Rate(int rating)
    {
        Console.WriteLine($"Фильм {Title} жанра {Genre} оценен на {rating}/10");
    }
}

public class AnimatedMovie : Movie
{

}
```

```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    private string category;
    private string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    protected string Category
    {
        get { return category; }
        set { category = value; }
    }

    internal string Description
    {
        get { return description; }
        set { description = value; }
    }
    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = Quantity + amount;
    }

    public string GetProductInfo()
    {
        string info = $"Товар: {Name}, Цена: {Price}, Количество: {Quantity}, Категория: {Category}";
        return info;
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    private decimal salary;
    private string contactInfo;
    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    protected decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    internal string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар \"{product.Name}\" добавлен продавцу {Name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продано {quantity} ед. товара \"{product.Name}\". Остаток: {product.Quantity}.");
    }

    public string GetSellerInfo()
    {
        string info = $"Продавец: {Name}, ID: {EmployeeId}, Контакты: {ContactInfo}";
        return info;
    }
}

public class Store
{
    private string storeName;
    private string location;
    private string storeHours;
    private List<Seller> sellers = new List<Seller>();
    private List<Product> productsInStore = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин \"{StoreName}\".");
    }

    public void AddProductToStore(Product product)
    {
        productsInStore.Add(product);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине \"{StoreName}\":");
        foreach (Product product in productsInStore)
        {
            Console.WriteLine(" - " + product.GetProductInfo());
        }
    }

    public string GetStoreInfo()
    {
        string info = $"Магазин: {StoreName}, Адрес: {Location}, Часы работы: {StoreHours}";
        return info;
    }
}

Product product1 = new Product("Ноутбук", 55000m, 10, "Электроника", "Игровой ноутбук");
Product product2 = new Product("Мышь", 1500m, 30, "Аксессуары", "Беспроводная мышь");

Seller seller1 = new Seller("Дмитрий", "EMP001", 60000m, "Dima2000@mail.com");

Store store1 = new Store("ТехноМир", "ул. Ленина, 10", "10:00 - 21:00");

store1.AddSeller(seller1);

seller1.AddProduct(product1);
seller1.AddProduct(product2);

store1.AddProductToStore(product1);
store1.AddProductToStore(product2);

store1.ListProducts();

seller1.SellProduct(product1, 3);
seller1.SellProduct(product2, 5);

Console.WriteLine();
Console.WriteLine(product1.GetProductInfo());
Console.WriteLine(product2.GetProductInfo());
Console.WriteLine(seller1.GetSellerInfo());
Console.WriteLine(store1.GetStoreInfo());

Console.WriteLine($"Общая стоимост : {product1.GetTotalPrice()}");
Console.WriteLine($"Общая стоимость: {product2.GetTotalPrice()}");
```
