```
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый класс с дополнительными атрибутами
public class Animal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public Animal(string name, string species, int age, double weight)
    {
        Name = name;
        Species = species;
        Age = age;
        Weight = weight;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"[{Species}] {Name}, Возраст: {Age}, Вес: {Weight}кг");
    }

    // Событие: животное покормили
    public delegate void AnimalFedHandler(string message);
    public event AnimalFedHandler OnFed;

    public void Feed(string food)
    {
        OnFed?.Invoke($"{Name} ({Species}) был(а) покормлен(а) {food}.");
    }
}

// Производные классы с уникальными методами и свойствами
public class Dog : Animal
{
    public string Breed { get; set; }
    
    public Dog(string name, string breed, int age, double weight) : base(name, "Собака", age, weight) 
    { 
        Breed = breed; 
    }
    
    public void Bark() => Console.WriteLine($"{Name} говорит Гав!");
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, double weight, bool isIndoor) : base(name, "Кошка", age, weight) 
    { 
        IsIndoor = isIndoor; 
    }
    
    public void Meow() => Console.WriteLine($"{Name} говорит Мяу!");
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double weight, bool canFly) : base(name, "Птица", age, weight) 
    { 
        CanFly = canFly; 
    }
    
    public void Sing() => Console.WriteLine($"{Name} красиво поет!");
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }
}

// --- Демонстрация работы проекта ---

// 1. Коллекция
List<Animal> zoo = new List<Animal>
{
    new Dog("Рекс", "Немецкая овчарка", 3, 30.5),
    new Cat("Луна", 2, 4.2, true),
    new Bird("Скай", 1, 0.5, true),
    new Dog("Бадди", "Золотистый ретривер", 5, 35.0)
};

// 2. Лямбда-выражения (фильтрация животных старше 2 лет)
var olderAnimals = zoo.Where(a => a.Age > 2).ToList();
Console.WriteLine("--- Животные старше 2 лет ---");
olderAnimals.ForEach(a => a.DisplayInfo());

// 3. Делегаты (вывод всех животных через Action)
Action<Animal> displayAction = a => {
    Console.WriteLine();
    a.DisplayInfo();
};
Console.WriteLine("\n--- Вывод всех через делегат ---");
zoo.ForEach(displayAction);

// 4. События (подписка и вызов)
Console.WriteLine("\n--- Демонстрация события ---");
zoo[0].OnFed += (message) => Console.WriteLine($"[Лог события]: {message}");
zoo[0].Feed("премиальным кормом для собак");
```
