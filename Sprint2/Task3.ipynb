```
Documentary doc = new Documentary("Планета Земля", "Алистер Фотергилл", 2006, 50, "Великобритания", "Природа", "Дэвид Аттенборо", "разные континенты");
HorrorFilm horror = new HorrorFilm("Оно", "Андрес Мускетти", 2017, 135, "США", "Ужасы", "Билл Скарсгард", 35, 8, 16, true);
AnimatedMovie cartoon = new AnimatedMovie("Тайна Коко", "Ли Анкрич", 2017, 105, "США", "Pixar", "3D", 6, "Disney+");

Viewer anna = new Viewer("Анна");

Console.WriteLine("=== Простое наследование: Movie -> Documentary ===");
anna.WatchAndRate(doc, 7);
doc.ShowNarrator();
doc.GetLocationInfo();
Console.WriteLine(doc.GetInfo());
Console.WriteLine($"Возраст фильма: {doc.GetAge()} лет");

Console.WriteLine();
Console.WriteLine("=== Сложное наследование: Movie -> FeatureFilm -> HorrorFilm ===");
anna.WatchAndRate(horror, 9);
horror.ShowCast();
Console.WriteLine(horror.GetWarning());
Console.WriteLine(horror.GetInfo());
Console.WriteLine($"Длинный фильм: {horror.IsLong()}");
Console.WriteLine($"Блокбастер: {horror.IsBlockbuster()}");

Console.WriteLine();
Console.WriteLine("=== Множественное наследование: Movie + IStreamable + IAwardable ===");
anna.WatchAndRate(cartoon, 10);
cartoon.ShowTechnique();
cartoon.Stream();
cartoon.AddAward("Оскар");

IAwardable awardable = cartoon;   // работаем с объектом через интерфейс
awardable.AddAward("BAFTA");

Console.WriteLine(cartoon.GetInfo());

public interface IStreamable
{
    string Platform { get; set; }
    void Stream();
}

public interface IAwardable
{
    int AwardsCount { get; set; }
    void AddAward(string awardName);
}

public class Movie
{
    public string Title {get; set;}
    public string Director {get; set;}
    public int ReleaseYear {get; set;}

    public int Duration {get; set;}
    public string Country {get; set;}
    public int Rating {get; set;}

    public Movie(string title, string director, int releaseyear,int duration, string country)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseyear;
        Duration = duration;
        Country = country;
    }

    public virtual string GetInfo()
    {
        return $"{Title}, {Director}, {ReleaseYear}, {Duration} мин, {Country}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"Вы смотрите фильм: {Title}");
    }

    public virtual void Rate(int rating)
    {
        Rating = rating;
        Console.WriteLine($"{Title} оценен на {rating}/10");
    }

    public bool IsLong(){
        return Duration > 120;
    }
    public int GetAge(){
        return DateTime.Now.Year - ReleaseYear;
    }
}   

public class Documentary : Movie
{
    public string Theme {get; set;}
    public string Narrator {get; set;}
    public string FilmingLocation {get; set;}

    public Documentary(string title, string director, int releaseyear, int duration, string country, string theme, string narrator, string filmingLocation) 
    : base (title, director, releaseyear, duration, country)
    {
        Theme = theme;
        Narrator = narrator;
        FilmingLocation = filmingLocation;
    }

    public override void Watch()
    {
        Console.WriteLine($"Вы смотрите фильм {Title} на тему {Theme}");
    }

    public void ShowNarrator(){
        Console.WriteLine($"Фильм озвучил {Narrator}");
    }

    public void GetLocationInfo(){
        Console.WriteLine($"Снимали в: {FilmingLocation}");
    }
}

public class FeatureFilm : Movie
{
    public string Genre {get; set;}
    public string LeadActor {get; set;}
    public double Budget {get; set;}

    public FeatureFilm(string title, string director, int releaseyear, int duration, string country, string genre, string leadActor, double budget)
    : base(title, director, releaseyear, duration, country)
    {
        Genre = genre;
        LeadActor = leadActor;
        Budget = budget;
    }

    public override void Rate(int rating){
        Rating = rating;

        Console.WriteLine($"Игровой фильм '{Title}' ({Genre}) оценен на {rating}/10");
        if (Genre.Equals("Ужасы", StringComparison.OrdinalIgnoreCase) && rating >= 8){
            Console.WriteLine($"Отличный хоррор, действительно держит в напряжении!");
        }
        else if (Genre.Equals("Комедия", StringComparison.OrdinalIgnoreCase) && rating >= 8){
            Console.WriteLine($"Замечательная комедия, отлично поднимает настроение!");
        }
    }

    public void ShowCast(){
        Console.WriteLine($"В главной роли: {LeadActor}");
    }
    public bool IsBlockbuster(){
            return Budget >= 100;
    }
}

public class HorrorFilm : FeatureFilm
{
    public int ScareLevel {get; set;}
    public int MinAge {get; set;}
    public bool HasJumpScares {get; set;}

    public HorrorFilm(string title, string director, int releaseyear, int duration, string country, string genre, string leadActor, double budget,
    int scareLevel, int minAge, bool hasJumpScares)
    : base(title, director, releaseyear, duration, country, "Ужасы", leadActor, budget){
        ScareLevel = scareLevel;
        MinAge = minAge;
        HasJumpScares = hasJumpScares;
    }

    public string GetWarning(){
        string scares = HasJumpScares ? "есть скримеры" : "без скримеров";
        return $"Возраст: {MinAge}+, уровень страха {ScareLevel}/10, {scares}";
    }
}


public class AnimatedMovie : Movie, IStreamable, IAwardable
{
    public string AnimationStudio {get; set;}
    public string Technique {get; set;}
    public int TargetAge {get; set;}
    public string Platform { get; set; }
    public int AwardsCount { get; set; }

    public AnimatedMovie(string title, string director, int releaseyear, int duration, string country, string animationStudio,
    string technique, int targetAge, string platform)
    :base(title, director, releaseyear, duration, country){
        AnimationStudio = animationStudio;
        Technique = technique;
        TargetAge = targetAge;
        Platform = platform;
        AwardsCount = 0;
    }

    public override string GetInfo(){
        return $"{base.GetInfo()} | Студия анимации: {AnimationStudio}";
    }

    public void ShowTechnique(){
        Console.WriteLine($"Техника анимации: {Technique}, с {TargetAge} лет");
    }

    public void Stream(){
        Console.WriteLine($"'{Title}' транслируется на платформе {Platform}");
    }

    public void AddAward(string awardName){
        AwardsCount++;
        Console.WriteLine($"'{Title}' получил награду: {awardName} (всего: {AwardsCount})");
    }
}

public class Viewer{
    public string Name {get; set;}
    public Viewer(string name){
        Name = name;
    }

    public void WatchAndRate(Movie movie, int rating){
        Console.WriteLine($"{Name} начинает просмотр");
        movie.Watch();
        movie.Rate(rating);
    }
}

```
